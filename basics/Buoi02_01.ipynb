**Câu 9.** Lọc ra các đơn hàng có mức giảm giá (`Discount`) từ 0.3 trở lên. Doanh số (`SALES`) trung bình của nhóm này là bao nhiêu, so với doanh số trung bình toàn bộ dữ liệu?

```python
mean_all = df['SALES'].mean()
mean_discount = df.loc[(df['Discount'] >= 0.3), 'SALES'].mean()
ratio = ((mean_discount - mean_all) / mean_all) * 100
print(f'Doanh số trung bình của nhóm này là: {mean_discount:.4f} ')
print(f'Doanh thu trung bình của nhóm này bằng {ratio:.4f} so với doanh thu trung bình của toàn bộ dữ liệu')
```

## Phần 3: Tạo cột mới và biến đổi dữ liệu


**Câu 10.** Từ cột `ORDERDATE`, tạo thêm 3 cột mới: `ORDER_YEAR`, `ORDER_MONTH`, `ORDER_QUARTER` chứa năm, tháng, quý của ngày đặt hàng.

```python
df['ORDER_YEAR'] = df['ORDERDATE'].dt.year
df['ORDER_MONTH'] = df['ORDERDATE'].dt.month
df['ORDER_QUARTER'] = df['ORDERDATE'].dt.quarter
df
```

**Câu 11.** Tạo cột mới `PROFIT_MARGIN` = `Profit` / `SALES` (tỷ suất lợi nhuận). Đơn hàng nào có tỷ suất lợi nhuận cao nhất? Thấp nhất?

```python
df['PROFIT_MARGIN'] = df['Profit'] / df['SALES']
result = df['PROFIT_MARGIN'].sort_values(ascending=False)
print(f'Đơn hàng có tỷ suất lợi nhuận cao nhất là đơn hàng số: {result.index[0]} với tỷ suất lợi nhuận là: {result.loc[result.index[0]]:.4f}')
print(f'Đơn hàng có tỷ suất lợi nhuận thấp nhất là đơn hàng số: {result.index[-1]} với tỷ suất lợi nhận là: {result.loc[result.index[-1]]:.4f}')
```

**Câu 12.** Tạo cột `NET_SALES` = `SALES` * (1 - `Discount`) thể hiện doanh số sau khi trừ giảm giá. So sánh tổng `SALES` và tổng `NET_SALES` của toàn bộ dữ liệu.

```python
df['NET_SALES'] = df['SALES'] * (1 - df['Discount'])
total_sales = df['SALES'].sum()
total_net_sales = df['NET_SALES'].sum()
diff = total_sales - total_net_sales
print(f'tổng sales là: {total_sales:.4f} và tổng net_sales là: {total_net_sales:.4f}')
print(f'khoảng chênh lệch giữa tổng sales và tổng net_sales là: {diff:.4f}')
```

## Phần 4: Nhóm dữ liệu (Groupby) và tổng hợp


**Câu 13.** Tính tổng `SALES` và tổng `Profit` theo từng `COUNTRY`. Sắp xếp giảm dần theo tổng `SALES` và cho biết Top 5 quốc gia có doanh số cao nhất.

```python
result = df.groupby('COUNTRY')[['SALES', 'Profit']].sum().sort_values(by='SALES', ascending=False)
result.head(5)
```

**Câu 14.** Tính `SALES` trung bình và `Profit` trung bình theo từng nhóm `DEALSIZE`. Nhận xét: nhóm `Large` có thực sự mang lại lợi nhuận trung bình cao nhất không?

```python
result = df.groupby('DEALSIZE')[['SALES', 'Profit']].mean().sort_values(by='Profit', ascending=False)
print(result)
print("large không mang lại lợi nhuận trung bình cao nhất")
```

**Câu 15.** Tính số lượng đơn hàng (đếm số `ORDERNUMBER` duy nhất) theo từng `STATUS`. Tỷ lệ đơn hàng bị `Cancelled` hoặc `Disputed` chiếm bao nhiêu % tổng số đơn hàng?

```python
ordernumber_count = df.groupby('STATUS')['ORDERNUMBER'].nunique()
total_ordernumber = df['ORDERNUMBER'].nunique()
target_count = ordernumber_count['Cancelled'] + ordernumber_count['Disputed']
target_percent = (target_count / total_ordernumber) * 100
print(target_percent.round(4))
```

**Câu 16.** Sử dụng `pivot_table` để tạo bảng tổng hợp tổng `SALES` theo `COUNTRY` (hàng) và `DEALSIZE` (cột).

```python
table = pd.pivot_table(df, index='COUNTRY', columns='DEALSIZE', values='SALES', aggfunc='sum')
table
```

**Câu 17.** Theo `ORDER_YEAR` đã tạo ở Câu 10, tính tổng doanh số (`SALES`) theo từng năm. Doanh số năm nào cao nhất? Lưu ý gì về dữ liệu năm 2005 (dữ liệu có đủ 12 tháng không)?

```python
result = df.groupby("ORDER_YEAR")['SALES'].sum().sort_values(ascending=False)
print(result)
print(f"doanh số năm: {result.index[0]} là cao nhất là: {result.iloc[0]}")
```
